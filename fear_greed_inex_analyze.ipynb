```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load the dataset
df_sentiment=pd.read_csv('/content/drive/MyDrive/fear_greed_index_2.csv')
```

```python
#convert 'date' to a proper datetime format
df_sentiment['timestamp']=pd.to_datetime(df_sentiment['timestamp'])
```

```python
#basic exploration
print('----Data Info')
print("df_sentiment.info()")
print("\n----first 5 rows----")
print(df_sentiment.head())
```

```python
# check the distribution of sentimate
#this tell you how often the market is in 'fear' vs 'greed'
sentiment_counts=df_sentiment["classification"].value_counts()
print("\n----sentiment counts----")
print(sentiment_counts)
```

```python
#visualize the Trend
plt.figure(figsize=(12,6))
plt.plot(df_sentiment["timestamp"],df_sentiment['value'],color="blue",label='sentiment value')
plt.title('Bitcoin Fear & Greed Index over Time ')
plt.xlabel('Date')
plt.ylabel('Index value (0-100)')
plt.grid(True)
plt.legend()
plt.show()
```
